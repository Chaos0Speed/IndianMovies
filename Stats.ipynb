```python
# Why: the only libraries the 'initial stats' checkpoint deliverable needs -- this
# notebook exists solely to produce Stats.csv for the first progress report.

import pandas as pd
import numpy as np
```

```python
# Why: describe() is the per-column summary (count / mean / std / quartiles) that was
# committed as the initial statistical snapshot of the cleaned data.

df = pd.read_csv('CleanedDataSet.csv')
stats = (df.describe())
```

```python
# Why: persists the summary so it can be cited without re-running; index=True keeps
# the column names as a column. NOTE: naming the variable `stats` is a shadowing hazard --
# if scipy is ever imported here (`from scipy import stats`) it would be overwritten.

stats.to_csv('Stats.csv', index=True)
```
